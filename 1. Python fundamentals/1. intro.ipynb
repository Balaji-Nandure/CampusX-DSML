```python
# data types
print(type("Hello World")) # string
print(type(5)) # integer
print(type(5.2)) # float
print(type(True))  # boolean
print(type(None)) # NoneType
print(type([1, 2, 3])) # list
print(type((1, 2, 3))) # tuple
print(type({"name": "Balaji", "age": 30})) # dictionary
print(type({1, 2, 3})) # set

```

```python
# variables
name = "Balaji"
age = 24
print(name)
print(age)
```

```python
# dynamic typing
# In Python, the type of a variable can change depending on the value you assign to it.

x = 5         # x is an integer
print(type(x))  # <class 'int'>

x = "Hello"   # x is now a string
print(type(x))  # <class 'str'>

x = 3.14      # x is now a float
print(type(x))  # <class 'float'>
```

```python
# multiple variables
a, b, c = 1, 2, 3 # a = 1, b = 2, c = 3
a = b = c = 1 # a = 1, b = 1, c = 1
```

```python
# taking input from user and adding them
a = int(input("Enter a number: "))
b = int(input("Enter another number: "))
print(a + b)
```

```python
# examples of explicit type conversion

# Converting string to int
num_str = "42"
num_int = int(num_str)
print(num_int, type(num_int))  # 42 <class 'int'>

# Converting float to int (note that the decimal part is dropped)
num_float = 3.99
num_int2 = int(num_float)
print(num_int2, type(num_int2))  # 3 <class 'int'>

# Converting int to string
num = 77
num_str2 = str(num)
print(num_str2, type(num_str2))  # '77' <class 'str'>

# Converting int to float
num3 = 7
num_float2 = float(num3)
print(num_float2, type(num_float2))  # 7.0 <class 'float'>

# Converting list to tuple
l = [1, 2, 3]
t = tuple(l)
print(t, type(t))  # (1, 2, 3) <class 'tuple'>
```

```python
print(5 / 3) # float division
print(5 // 3) # integer division
print(5 % 3) # modulus
print(5 ** 3) # exponentiation
```
