```python
print("Hello World")

#  print mulitple things
print("Hello", 5, 5.2)

# changing separator
print("Hello", 5, 5.2, sep="/")

# changing end
print("Hello", 5, 5.2, end="|||||||||||")

```

```python
# data types
print(type("Hello World")) # string
print(type(5)) # integer
print(type(5.2)) # float
print(type(True))  # boolean
print(type(None)) # NoneType
print(type([1, 2, 3])) # list
print(type((1, 2, 3))) # tuple
print(type({"name": "Balaji", "age": 30})) # dictionary
print(type({1, 2, 3})) # set

```

```python
# variables
name = "Balaji"
age = 24
print(name)
print(age)
```

```python
# dynamic typing
# In Python, the type of a variable can change depending on the value you assign to it.

x = 5         # x is an integer
print(type(x))  # <class 'int'>

x = "Hello"   # x is now a string
print(type(x))  # <class 'str'>

x = 3.14      # x is now a float
print(type(x))  # <class 'float'>
```

```python
# multiple variables
a, b, c = 1, 2, 3 # a = 1, b = 2, c = 3
a = b = c = 1 # a = 1, b = 1, c = 1
```

```python
# taking input from user and adding them
a = int(input("Enter a number: "))
b = int(input("Enter another number: "))
print(a + b)

```
