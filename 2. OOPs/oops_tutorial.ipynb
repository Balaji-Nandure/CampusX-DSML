## 1. Core Concepts: Classes & Objects

*   **Class**: A blueprint or template for creating objects. It defines a set of attributes (data) and methods (behavior) that the created objects will have.
*   **Object**: An instance of a class. It is a concrete entity that has state and behavior.
*   **`self`**: Represents the specific instance of the class. It allows access to the attributes and methods of that instance.
*   **`__init__`**: The constructor method. It is automatically called when a new object of the class is instantiated.


```python
class Smartphone:
    # Class attribute (shared by all instances)
    platform = "Android"

    # Constructor method
    def __init__(self, brand, model, price):
        # Instance attributes (unique to each instance)
        self.brand = brand
        self.model = model
        self.price = price
        
    # Instance method
    def get_details(self):
        return f"Brand: {self.brand}, Model: {self.model}, Price: ${self.price}, OS: {self.platform}"

# Instantiating objects (Creating instances of the Smartphone class)
phone1 = Smartphone("Samsung", "Galaxy S24", 999)
phone2 = Smartphone("Google", "Pixel 8 Pro", 899)

# Accessing attributes and methods
print(phone1.get_details())
print(phone2.get_details())

# Accessing class attributes
print(f"Phone 1 platform: {phone1.platform}")
print(f"Phone 2 platform: {phone2.platform}")
```

### Key Takeaways:
- Class attributes (like `platform`) are shared across all instances. Modifying them affects all instances.
- Instance attributes (like `brand`, `model`, `price`) are bound to individual objects.
- `self` acts as a placeholder for the object that is calling the method.


## 2. Instance, Class, and Static Methods

Python supports three types of methods inside a class:
1.  **Instance Methods**: Take `self` as the first argument. They can modify object state and access class attributes.
2.  **Class Methods**: Take `cls` as the first argument and are marked with `@classmethod`. They can modify class state but not instance state. Often used as alternative constructors (factory methods).
3.  **Static Methods**: Marked with `@staticmethod`. They do not take `self` or `cls`. They behave like regular functions but reside inside the class's namespace for organizational utility.


```python
class Employee:
    # Class attribute
    raise_amount = 1.05  # 5% annual raise
    
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    # 1. Instance Method
    def apply_raise(self):
        self.salary = int(self.salary * self.raise_amount)
        return self.salary

    # 2. Class Method
    @classmethod
    def set_raise_amt(cls, amount):
        cls.raise_amount = amount

    # 2b. Class Method as a Factory (creates an object from a string)
    @classmethod
    def from_string(cls, emp_str):
        # Expecting format 'Name-Salary'
        name, salary = emp_str.split('-')
        return cls(name, int(salary))

    # 3. Static Method (independent utility function)
    @staticmethod
    def is_workday(day):
        # Monday = 0, Friday = 4, Saturday = 5, Sunday = 6
        return day.weekday() < 5


# --- Demonstration ---
import datetime

# Using factory classmethod to create an employee
emp_str_1 = "Alice-70000"
emp_1 = Employee.from_string(emp_str_1)
print(f"Created via factory: {emp_1.name}, Salary: ${emp_1.salary}")

# Using standard instance method
print(f"Salary before raise: ${emp_1.salary}")
emp_1.apply_raise()
print(f"Salary after raise: ${emp_1.salary}")

# Changing class attribute using classmethod
print(f"Old raise amount: {Employee.raise_amount}")
Employee.set_raise_amt(1.10)  # Change raise to 10%
print(f"New raise amount: {Employee.raise_amount}")

# Using static method
my_date = datetime.date(2026, 5, 25) # May 25, 2026 (Monday)
print(f"Is {my_date} a workday? {Employee.is_workday(my_date)}")
```

### Pillar 1: Encapsulation

**Encapsulation** is the practice of keeping fields (attributes) and methods private, and exposing them only through public interfaces. This protects data from unauthorized access and modifications.

In Python, access specifiers are implemented using naming conventions:
- **Public**: Accessible from anywhere (`name`).
- **Protected**: Indicated by a single underscore (`_name`). It is a convention warning other programmers not to access it outside the class or subclasses.
- **Private**: Indicated by double underscores (`__name`). Python performs **name mangling** on private variables to make them harder to access directly from outside the class.

To access and modify private data safely, we use **Getters and Setters**. Python provides a clean way to do this using `@property`.


```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner          # Public attribute
        self._account_type = "Checkings" # Protected attribute
        self.__balance = balance    # Private attribute (Name mangled to _BankAccount__balance)

    # Getter property for balance
    @property
    def balance(self):
        print("Retrieving account balance safely...")
        return self.__balance

    # Setter property for balance
    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            print("Updating account balance...")
            self.__balance = amount
        else:
            print("Error: Balance cannot be negative!")

    # Public method displaying encapsulation
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Deposit amount must be positive.")


# --- Demonstration ---
account = BankAccount("John Doe", 5000)

# Accessing Public Variable
print(f"Account Owner: {account.owner}")

# Trying to access Private Variable directly will raise an AttributeError
try:
    print(account.__balance)
except AttributeError as e:
    print(f"Expected Error: {e}")

# Accessing using Getter (@property)
print(f"Balance: ${account.balance}")

# Updating using Setter (@balance.setter)
account.balance = 6000
print(f"New Balance: ${account.balance}")

# Trying to set an invalid balance
account.balance = -100
print(f"Balance after invalid update: ${account.balance}")
```

### Pillar 2: Inheritance

**Inheritance** allows a new class (Subclass/Child) to adopt the attributes and methods of an existing class (Superclass/Parent). This promotes code reusability.

#### Types of Inheritance:
1.  **Single Inheritance**: One child class inherits from one parent class.
2.  **Multilevel Inheritance**: A child class inherits from a parent class, which inherits from another parent class (Grandparent -> Parent -> Child).
3.  **Multiple Inheritance**: A child class inherits from more than one parent class.
4.  **Hierarchical Inheritance**: Multiple child classes inherit from a single parent class.

#### Key Functions & Concepts:
-   **`super()`**: Allows us to call methods and constructors of the parent class.
-   **Method Overriding**: When a child class provides a specific implementation of a method that is already provided by its parent class.


```python
# Parent Class (Superclass)
class Vehicle:
    def __init__(self, brand, year):
        self.brand = brand
        self.year = year

    def start_engine(self):
        return "Engine started. Vroom!"

# Child Class (Subclass) - Single Inheritance
class Car(Vehicle):
    def __init__(self, brand, year, doors):
        # Call the parent constructor using super()
        super().__init__(brand, year)
        self.doors = doors

    # Method Overriding: Changing parent method's behavior for Car
    def start_engine(self):
        return f"The {self.brand} car's engine purrs smoothly."

# Multilevel Inheritance
class ElectricCar(Car):
    def __init__(self, brand, year, doors, battery_capacity):
        super().__init__(brand, year, doors)
        self.battery_capacity = battery_capacity

    # Overriding again
    def start_engine(self):
        return f"The {self.brand} electric car hums silently."


# --- Multiple Inheritance and MRO ---
class Engine:
    def get_engine_type(self):
        return "Internal Combustion"

class ElectricMotor:
    def get_motor_type(self):
        return "AC Induction Motor"

# Hybrid inherits from two unrelated classes
class HybridCar(Vehicle, Engine):
    def __init__(self, brand, year):
        Vehicle.__init__(self, brand, year)


# --- Demonstration ---
standard_car = Car("Ford", 2023, 4)
electric_car = ElectricCar("Tesla", 2024, 4, "85 kWh")
hybrid_car = HybridCar("Toyota", 2022)

print(standard_car.start_engine())
print(electric_car.start_engine())
print(f"Hybrid car brand: {hybrid_car.brand}, Engine: {hybrid_car.get_engine_type()}")

# Check Method Resolution Order (MRO) for HybridCar
print(f"MRO for HybridCar: {HybridCar.__mro__}")
```

### Pillar 3: Polymorphism

**Polymorphism** (meaning "many forms") is the ability to present the same interface for different underlying forms (data types or classes).

In Python, polymorphism is demonstrated via:
1.  **Duck Typing**: Dynamic typing concept where an object's suitability is determined by the presence of certain methods and properties, rather than its inheritance pedigree. ("If it walks like a duck and quacks like a duck, it is a duck.")
2.  **Method Overriding**: Subclasses overriding methods of the parent class (seen in the Inheritance section).
3.  **Operator Overloading**: Giving new meaning to operators (like `+`, `*`, `==`) using magic/dunder methods.


```python
# --- 1. Duck Typing ---
class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

class Robot:
    def speak(self):
        return "Beep Boop!"

# This function takes any object that has a 'speak' method, regardless of class hierarchy
def make_it_speak(animal_or_object):
    print(animal_or_object.speak())

dog = Dog()
cat = Cat()
robot = Robot()

print("--- Duck Typing Demonstration ---")
make_it_speak(dog)
make_it_speak(cat)
make_it_speak(robot)


# --- 2. Operator Overloading ---
class Point2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the '+' operator
    def __add__(self, other):
        return Point2D(self.x + other.x, self.y + other.y)

    # Overloading the string representation
    def __str__(self):
        return f"({self.x}, {self.y})"

p1 = Point2D(2, 3)
p2 = Point2D(4, 5)
p3 = p1 + p2  # Internally calls p1.__add__(p2)

print("\n--- Operator Overloading Demonstration ---")
print(f"Point 1: {p1}")
print(f"Point 2: {p2}")
print(f"Point 1 + Point 2 = {p3}")
```

### Pillar 4: Abstraction

**Abstraction** is the process of hiding internal details and showing only the essential features of an object. It acts as a contract that subclasses must follow.

In Python, abstraction is achieved using the `abc` (Abstract Base Classes) module:
- Inherit from `ABC`.
- Decorate abstract methods with `@abstractmethod`.
- **Note**: You *cannot* instantiate an abstract class. Any subclass inheriting from it *must* implement all its abstract methods, or it will also remain abstract.


```python
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    
    @abstractmethod
    def validate_payment(self, amount):
        """Validates payment details."""
        pass

    @abstractmethod
    def process_payment(self, amount):
        """Processes the payment transactions."""
        pass

class PayPalPayment(PaymentProcessor):
    def validate_payment(self, amount):
        print(f"PayPal: Validating transaction of ${amount}...")
        return True

    def process_payment(self, amount):
        if self.validate_payment(amount):
            print(f"PayPal: Successfully processed ${amount}.")

class StripePayment(PaymentProcessor):
    def validate_payment(self, amount):
        print(f"Stripe: Checking card details for ${amount}...")
        return True

    def process_payment(self, amount):
        if self.validate_payment(amount):
            print(f"Stripe: Card charged for ${amount}.")

# --- Demonstration ---
# Trying to instantiate abstract class raises error
try:
    processor = PaymentProcessor()
except TypeError as e:
    print(f"Expected Error: {e}")

# Instantiating concrete implementations
paypal = PayPalPayment()
stripe = StripePayment()

paypal.process_payment(250)
stripe.process_payment(450)
```

## 4. Magic (Dunder) Methods

Magic methods (or **Double Underscore / Dunder** methods) are special methods with double underscores at the beginning and end. They allow you to define how your custom classes behave with Python's built-in syntax (operators, print statements, length checking, etc.).

Common Dunder Methods:
- `__init__`: Constructor.
- `__str__`: User-friendly string representation (called by `print()` and `str()`).
- `__repr__`: Official/Developer-friendly string representation (called by debugging/interactive shell).
- `__len__`: Custom length behaviour (called by `len()`).
- `__eq__`: Custom equality behaviour (called by `==`).


```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    # 1. User-friendly string representation
    def __str__(self):
        return f"'{self.title}' by {self.author}"

    # 2. Unambiguous developer representation
    def __repr__(self):
        return f"Book(title='{self.title}', author='{self.author}', pages={self.pages})"

    # 3. Custom length behavior
    def __len__(self):
        return self.pages

    # 4. Custom equality comparison
    def __eq__(self, other):
        if isinstance(other, Book):
            return self.title == other.title and self.author == other.author
        return False


# --- Demonstration ---
book1 = Book("Python OOP Deep Dive", "John Smith", 350)
book2 = Book("Python OOP Deep Dive", "John Smith", 350)
book3 = Book("Learning Algorithms", "Jane Doe", 420)

print(f"__str__ call (via print): {book1}")
print(f"__repr__ call: {repr(book1)}")
print(f"__len__ call: {len(book1)} pages")
print(f"Is book1 == book2? {book1 == book2}")
print(f"Is book1 == book3? {book1 == book3}")
```

## 5. Real-World Project: Smart Banking System

Let's integrate all 4 pillars and advanced OOP concepts into a complete, real-world application: a **Smart Banking and Portfolio System**.

This system utilizes:
1.  **Abstraction**: `Account` as an abstract base class.
2.  **Encapsulation**: Protected account details and private balance with property getters and setters.
3.  **Inheritance**: `SavingsAccount` and `InvestmentAccount` inheriting from `Account`.
4.  **Polymorphism**: Overriding interest calculations and withdrawal mechanics, plus duck typing for general statement printing.
5.  **Dunder Methods**: Custom string formatting and arithmetic addition of multiple accounts.


```python
from abc import ABC, abstractmethod

# Abstract Base Class representing Abstraction
class Account(ABC):
    def __init__(self, account_holder, account_number, balance=0.0):
        self.account_holder = account_holder
        self._account_number = account_number  # Protected
        self.__balance = balance                # Private

    # Encapsulation: Getter for Private Balance
    @property
    def balance(self):
        return self.__balance

    # Encapsulation: Setter with validation
    @balance.setter
    def balance(self, new_balance):
        if new_balance >= 0:
            self.__balance = new_balance
        else:
            raise ValueError("Balance cannot be negative!")

    @property
    def account_number(self):
        return self._account_number

    # Encapsulated deposit method
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount:.2f} into Account {self.account_number}.")
        else:
            print("Deposit amount must be positive.")

    # Abstract methods that must be implemented by subclasses
    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def calculate_interest(self):
        pass


# Subclass 1: Inheritance and Polymorphism
class SavingsAccount(Account):
    def __init__(self, account_holder, account_number, balance=0.0, interest_rate=0.02):
        super().__init__(account_holder, account_number, balance)
        self.interest_rate = interest_rate

    def withdraw(self, amount):
        # Specific withdrawal rule (minimum balance check)
        min_balance = 100.0
        if self.balance - amount >= min_balance:
            self.balance -= amount
            print(f"Withdrew ${amount:.2f} from Account {self.account_number}. Current Balance: ${self.balance:.2f}")
        else:
            print(f"Transaction Denied! Must maintain a minimum balance of ${min_balance:.2f}.")

    def calculate_interest(self):
        earned = self.balance * self.interest_rate
        self.balance += earned
        print(f"Applied {self.interest_rate*100}% interest. Earned: ${earned:.2f}. New Balance: ${self.balance:.2f}")


# Subclass 2: Inheritance and Polymorphism
class InvestmentAccount(Account):
    def __init__(self, account_holder, account_number, balance=0.0, risk_factor="Medium"):
        super().__init__(account_holder, account_number, balance)
        self.risk_factor = risk_factor

    def withdraw(self, amount):
        # Investment withdrawal has a fee
        withdrawal_fee = 25.0
        total_deduction = amount + withdrawal_fee
        if self.balance >= total_deduction:
            self.balance -= total_deduction
            print(f"Withdrew ${amount:.2f} (Fee: ${withdrawal_fee:.2f}) from Account {self.account_number}. Balance: ${self.balance:.2f}")
        else:
            print("Transaction Denied! Insufficient funds to cover withdrawal and transaction fees.")

    def calculate_interest(self):
        # Performance based on risk factor
        performance_map = {"Low": 0.04, "Medium": 0.08, "High": 0.15}
        rate = performance_map.get(self.risk_factor, 0.05)
        earned = self.balance * rate
        self.balance += earned
        print(f"Investment return ({self.risk_factor} risk: {rate*100}%): ${earned:.2f}. New Balance: ${self.balance:.2f}")

    # Magic / Dunder method to combine two accounts' balances
    def __add__(self, other):
        if isinstance(other, Account):
            return self.balance + other.balance
        return NotImplemented

    # User friendly representation
    def __str__(self):
        return f"InvestmentAccount({self.account_holder}, AccNo: {self.account_number}, Bal: ${self.balance:.2f})"


# --- Duck Typing Helper ---
def print_account_summary(account_obj):
    # This works for any account object because they share the same interface
    print(f"--- Account Summary ---")
    print(f"Holder: {account_obj.account_holder}")
    print(f"Number: {account_obj.account_number}")
    print(f"Balance: ${account_obj.balance:.2f}")
    print("-----------------------")


# --- Execution and Verification ---

# Create Savings and Investment Accounts
savings = SavingsAccount("Alice Green", "SAV-101", 500.0, interest_rate=0.03)
investment = InvestmentAccount("Alice Green", "INV-202", 2000.0, risk_factor="High")

# Demonstrate Polymorphism via Duck Typing
print_account_summary(savings)
print_account_summary(investment)

# Demonstrate Account Specific Operations (Polymorphic methods)
print("\n--- Running Financial Activities ---")
savings.deposit(100.0)
savings.withdraw(450.0)   # Should pass
savings.withdraw(100.0)   # Should fail minimum balance check

investment.withdraw(200.0) # Should charge fee
savings.calculate_interest()
investment.calculate_interest()

# Demonstrate Dunder Method (__add__)
total_worth = investment + savings
print(f"\nTotal combined wealth of Alice Green: ${total_worth:.2f}")

# Demonstrate Dunder Method (__str__)
print(investment)
```
